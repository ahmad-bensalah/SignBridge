## 5. Load Processor (Feature Extractor + Tokenizer)

```python
from transformers import WhisperProcessor, WhisperFeatureExtractor, WhisperTokenizer

processor = WhisperProcessor.from_pretrained(
    MODEL_NAME,
    language=LANGUAGE,
    task=TASK
)

feature_extractor = processor.feature_extractor
tokenizer         = processor.tokenizer

print("Processor loaded")
print(f"Vocab size    : {tokenizer.vocab_size}")
print(f"Sampling rate : {feature_extractor.sampling_rate}")
```

## 6. Preprocessing — Audio → Features + Labels

```python
import re

def clean_text(text: str) -> str:
    """Light normalisation — strip extra spaces."""
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    return text

def prepare_dataset(batch):
    # Load and resample audio
    audio = batch[AUDIO_COLUMN]

    # Compute log-Mel spectrogram features
    batch["input_features"] = feature_extractor(
        audio["array"],
        sampling_rate=audio["sampling_rate"]
    ).input_features[0]

    # Encode target text to token IDs
    batch["labels"] = tokenizer(clean_text(batch[TEXT_COLUMN])).input_ids
    return batch

print("Processing training set...")
train_dataset = train_dataset.map(
    prepare_dataset,
    remove_columns=train_dataset.column_names,
    num_proc=1
)

print("Processing eval set...")
eval_dataset = eval_dataset.map(
    prepare_dataset,
    remove_columns=eval_dataset.column_names,
    num_proc=1
)

print(f"✅ Preprocessing done. Sample input_features shape: {train_dataset[0]['input_features'].__class__}")
```

## 7. Data Collator

```python
from dataclasses import dataclass
from typing import Any, Dict, List, Union
import torch

@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    """
    Custom collator that:
      - Pads input_features to the same length (right-pad to 3000 frames)
      - Pads label sequences and replaces padding token id with -100
        so they are ignored by the cross-entropy loss
    """
    processor: Any

    def __call__(self, features: List[Dict[str, Any]]) -> Dict[str, torch.Tensor]:
        # ── Audio features ──
        input_features = [
            {"input_features": f["input_features"]} for f in features
        ]
        batch = self.processor.feature_extractor.pad(
            input_features, return_tensors="pt"
        )

        # ── Labels (token ids) ──
        label_features = [{"input_ids": f["labels"]} for f in features]
        labels_batch = self.processor.tokenizer.pad(
            label_features, return_tensors="pt"
        )

        # Replace padding token id with -100 → ignored in loss
        labels = labels_batch["input_ids"].masked_fill(
            labels_batch.attention_mask.ne(1), -100
        )

        # Remove BOS token added by tokenizer (Whisper adds it internally)
        if (labels[:, 0] == self.processor.tokenizer.bos_token_id).all():
            labels = labels[:, 1:]

        batch["labels"] = labels
        return batch


data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor=processor)
print("Data collator ready")
```

## 8. Evaluation Metric — WER

```python
import evaluate
import numpy as np

wer_metric = evaluate.load("wer")

def compute_metrics(pred):
    pred_ids   = pred.predictions
    label_ids  = pred.label_ids

    # Replace -100 back to pad_token_id so the tokenizer can decode them
    label_ids[label_ids == -100] = tokenizer.pad_token_id

    pred_str  = tokenizer.batch_decode(pred_ids,  skip_special_tokens=True)
    label_str = tokenizer.batch_decode(label_ids, skip_special_tokens=True)

    wer = 100 * wer_metric.compute(predictions=pred_str, references=label_str)
    return {"wer": wer}

print("WER metric loaded")
```

## 9. Load Whisper Model

```python
from transformers import WhisperForConditionalGeneration

model = WhisperForConditionalGeneration.from_pretrained(MODEL_NAME)

# Force decoder to use Arabic language + transcribe task
model.generation_config.language = LANGUAGE
model.generation_config.task = TASK
model.generation_config.forced_decoder_ids = None  # handled by generation_config

# Count parameters
total_params = sum(p.numel() for p in model.parameters()) / 1e6
trainable    = sum(p.numel() for p in model.parameters() if p.requires_grad) / 1e6

print(f"Model loaded: {MODEL_NAME}")
print(f"Total params    : {total_params:.1f}M")
print(f"Trainable params: {trainable:.1f}M")
```

## 10. Training Arguments

```python
from transformers import Seq2SeqTrainingArguments

training_args = Seq2SeqTrainingArguments(
    output_dir=OUTPUT_DIR,

    # ── Batch & gradient ──
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    gradient_accumulation_steps=GRAD_ACCUM_STEPS,
    gradient_checkpointing=True,          # saves VRAM at cost of ~20% speed

    # ── Optimizer & schedule ──
    learning_rate=LEARNING_RATE,
    warmup_steps=WARMUP_STEPS,
    max_steps=MAX_STEPS,
    optim="adamw_torch",
    lr_scheduler_type="linear",

    # ── Eval & save ──
    eval_strategy="steps",
    eval_steps=EVAL_STEPS,
    save_strategy="steps",
    save_steps=SAVE_STEPS,
    save_total_limit=2,
    load_best_model_at_end=True,
    metric_for_best_model="wer",
    greater_is_better=False,

    # ── Generation during eval ──
    predict_with_generate=True,
    generation_max_length=225,

    # ── Misc ──
    fp16=torch.cuda.is_available(),       # use FP16 on GPU
    dataloader_num_workers=2,
    logging_steps=25,
    report_to="none",
    push_to_hub=False,
)

print("Training arguments configured")
print(f"Effective batch size: {BATCH_SIZE * GRAD_ACCUM_STEPS}")
print(f"FP16 enabled        : {training_args.fp16}")
```

## 11. Training

```python
from transformers import Seq2SeqTrainer

max_label_len = model.config.max_target_positions
print(f"Whisper max target length: {max_label_len}")

# 1) Filter samples that are too long
if "labels" in train_dataset.column_names:
    print("Detected preprocessed dataset with 'labels'.")

    def is_short_enough_preprocessed(example):
        return len(example["labels"]) <= max_label_len

    train_dataset = train_dataset.filter(is_short_enough_preprocessed)
    eval_dataset  = eval_dataset.filter(is_short_enough_preprocessed)

else:
    print(f"Detected raw dataset with text column '{TEXT_COLUMN}'.")

    def is_short_enough_raw(example):
        text = str(example[TEXT_COLUMN]).strip()
        token_ids = processor.tokenizer(text).input_ids
        return len(token_ids) <= max_label_len

    train_dataset = train_dataset.filter(is_short_enough_raw)
    eval_dataset  = eval_dataset.filter(is_short_enough_raw)

    # 2) Convert raw dataset -> Whisper features + labels
    def prepare_dataset(batch):
        audio = batch[AUDIO_COLUMN]

        batch["input_features"] = processor.feature_extractor(
            audio["array"],
            sampling_rate=audio["sampling_rate"]
        ).input_features[0]

        batch["labels"] = processor.tokenizer(
            str(batch[TEXT_COLUMN]).strip(),
            max_length=max_label_len,
            truncation=True
        ).input_ids

        return batch

    train_dataset = train_dataset.map(
        prepare_dataset,
        remove_columns=train_dataset.column_names,
        desc="Preparing filtered train dataset"
    )

    eval_dataset = eval_dataset.map(
        prepare_dataset,
        remove_columns=eval_dataset.column_names,
        desc="Preparing filtered eval dataset"
    )

print(f"Train samples after filtering: {len(train_dataset)}")
print(f"Eval  samples after filtering: {len(eval_dataset)}")

# Sanity check
if len(train_dataset) > 0:
    if "labels" in train_dataset.column_names:
        max_train_len = max(len(x) for x in train_dataset["labels"])
        print(f"Max train label length after filtering: {max_train_len}")

# 3) Build trainer
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
    processing_class=processor,
)

# 4) Train
print("Starting training...")
train_result = trainer.train()

print("\nTraining complete!")
print(f"Total steps      : {train_result.global_step}")
print(f"Training loss    : {train_result.training_loss:.4f}")
print(f"Training runtime : {train_result.metrics['train_runtime']:.0f}s")
```

## 15. Profanity Filter (Tunisian Arabic)

🚫 BEWARE this section contains bad that are crucial to test the filtering system

Bad words are derived from the **T-HSAB** hate-speech dataset by extracting
words that appear predominantly in hate-labeled sentences.
All matched words are replaced with `****`.

```python
import re
import math
import openpyxl
from collections import Counter

XLSX_PATH = "/content/T-HSAB.xlsx"

AR_STOPWORDS = {
    'في', 'من', 'على', 'الى', 'إلى', 'عن', 'مع', 'هذا', 'هذه', 'ذلك', 'تلك',
    'هو', 'هي', 'هم', 'هن', 'انا', 'أنت', 'انتي', 'احنا', 'نحن', 'كان', 'كانت',
    'يكون', 'ما', 'لا', 'لم', 'لن', 'ثم', 'او', 'أو', 'بل', 'لكن', 'لأن', 'لان',
    'اذا', 'إذا', 'كل', 'بعض', 'أي', 'اي', 'حتى', 'مثل', 'بعد', 'قبل', 'عند',
    'بين', 'بس', 'برشا', 'شوية', 'كيف', 'علاش', 'شنو', 'شنية', 'شنوا', 'توا',
    'يا', 'و', 'ف', 'ب', 'ل', 'ك', 'اللي', 'الذي', 'التي', 'الذين', 'اللاتي',
    'الله', 'الحمد', 'لله', 'ربي', 'ناس', 'شعب', 'بلاد', 'بلد', 'تونس'
}

def normalize_arabic(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = text.strip().lower()
    text = re.sub(r'[\u0640]', '', text)
    text = re.sub(r'[\u0617-\u061A\u064B-\u0652\u0670]', '', text)
    text = re.sub(r'[أإآٱ]', 'ا', text)
    text = re.sub(r'ى', 'ي', text)
    text = re.sub(r'ؤ', 'و', text)
    text = re.sub(r'ئ', 'ي', text)
    text = re.sub(r'ة', 'ه', text)
    text = re.sub(r'[^\u0600-\u06FF0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def tokenize_arabic(text: str):
    text = normalize_arabic(text)
    return re.findall(r'[\u0600-\u06FF]{2,}', text)

def generate_ngrams(tokens, n_values=(1, 2, 3)):
    out = []
    for n in n_values:
        for i in range(len(tokens) - n + 1):
            out.append(' '.join(tokens[i:i+n]))
    return out

def is_hate_label(label: str) -> bool:
    if not isinstance(label, str):
        return False
    label = label.strip().lower()
    return label in {"hate", "abusive", "offensive", "hs", "toxic"}

def load_t_hsab_rows(path):
    wb = openpyxl.load_workbook(path, read_only=True)
    ws = wb[wb.sheetnames[0]]
    rows = list(ws.iter_rows(values_only=True))

    data = []
    for row in rows:
        if row is None:
            continue
        text = row[0] if len(row) > 0 else None
        label = row[1] if len(row) > 1 else None
        if isinstance(text, str) and isinstance(label, str):
            data.append((text, label))

    if not data:
        raise ValueError("Could not read text/label rows from the xlsx file.")

    return data

def extract_bad_words_from_xlsx(
    path,
    min_count=2,
    min_hate_ratio=0.80,
    min_log_odds=1.2,
    max_words=300
):
    rows = load_t_hsab_rows(path)

    hate_counter = Counter()
    all_counter = Counter()

    hate_docs = 0
    total_docs = 0

    for text, label in rows:
        tokens = tokenize_arabic(text)
        if not tokens:
            continue

        ngrams = set(generate_ngrams(tokens, n_values=(1, 2, 3)))

        cleaned_ngrams = []
        for gram in ngrams:
            parts = gram.split()
            if all(p in AR_STOPWORDS for p in parts):
                continue
            cleaned_ngrams.append(gram)

        total_docs += 1
        for gram in cleaned_ngrams:
            all_counter[gram] += 1

        if is_hate_label(label):
            hate_docs += 1
            for gram in cleaned_ngrams:
                hate_counter[gram] += 1

    nonhate_docs = max(total_docs - hate_docs, 1)

    candidates = []
    for gram, total_count in all_counter.items():
        hate_count = hate_counter.get(gram, 0)
        nonhate_count = total_count - hate_count

        if total_count < min_count:
            continue

        hate_ratio = hate_count / total_count
        p_hate = (hate_count + 1) / (hate_docs + 2)
        p_nonhate = (nonhate_count + 1) / (nonhate_docs + 2)
        log_odds = math.log(p_hate / p_nonhate)

        if hate_ratio >= min_hate_ratio and log_odds >= min_log_odds:
            candidates.append((gram, total_count, hate_count, hate_ratio, log_odds))

    candidates.sort(key=lambda x: (x[3], x[4], x[2], len(x[0])), reverse=True)
    bad_words = [c[0] for c in candidates[:max_words]]
    return set(bad_words), candidates

BAD_WORDS, STATS = extract_bad_words_from_xlsx(
    XLSX_PATH,
    min_count=2,
    min_hate_ratio=0.80,
    min_log_odds=1.2,
    max_words=300
)

# Optional: remove obvious false positives automatically
BLACKLIST_FALSE_POSITIVES = {
    'الامارات', 'ليبي', 'رسولكم', 'بو'
}
BAD_WORDS = {w for w in BAD_WORDS if w not in BLACKLIST_FALSE_POSITIVES}

_sorted_bad = sorted(BAD_WORDS, key=len, reverse=True)

# IMPORTANT FIX:
# allow optional attached Arabic prefix letters like و/ف/ب/ل/ك before the bad word
# and keep the prefix in output
if _sorted_bad:
    _pattern = re.compile(
        r'(^|[^\u0600-\u06FF])'          # left boundary
        r'([وفبلك]?)'                    # optional attached prefix
        r'(' + '|'.join(re.escape(w) for w in _sorted_bad) + r')'
        r'(?![\u0600-\u06FF])',          # right boundary
        re.UNICODE
    )
else:
    _pattern = re.compile(r'$.')  # matches nothing

def filter_profanity(text: str, replacement: str = '****') -> str:
    """
    Normalize text then censor bad words.
    Keeps attached prefix like و / ف / ب / ل / ك
    Example: وقحبة -> و****
    """
    text = normalize_arabic(text)

    def repl(m):
        left = m.group(1)
        prefix = m.group(2)
        return f"{left}{prefix}{replacement}"

    return _pattern.sub(repl, text)

#  __        ___    ____  _   _ ___ _   _  ____
#  \ \      / / \  |  _ \| \ | |_ _| \ | |/ ___|
#   \ \ /\ / / _ \ | |_) |  \| || ||  \| | |  _
#    \ V  V / ___ \|  _ <| |\  || || |\  | |_| |
#     \_/\_/_/   \_\_| \_\_| \_|___|_| \_|\____|
#
#  ==========================================================
#  CONTENT WARNING
#  ----------------------------------------------------------
#  This cell contains offensive and explicit words
#  collected from Tunisian dialect text.
#  These terms may include insults, slurs, profanity,
#  sexual language, and abusive expressions.
#  They are included strictly for research, moderation,
#  dataset preparation, filtering, or NLP analysis.
#  Do not interpret their presence as endorsement,
#  recommendation, or acceptable everyday usage.
#  Handle outputs carefully, especially in demos,
#  shared notebooks, classroom settings, or reports.
#  Viewer discretion is advised.
#  ==========================================================
#  PURPOSE: LINGUISTIC / MODERATION ANALYSIS ONLY
#  ==========================================================
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX
#   XXXXXXXXXXXXXXXXXXXXXXXXXX BAD WORDS XXXXXXXXXXXXXXXXXXXXXXXXXX


test_cases = [
    'مرحبا كيفاش حالك اليوم',
    'هذا الكلب ما يستاهلش',
    'برشا نيك وقحبة في الكلام',
    'بنت القحبة هاذي',
    'روح نيك امك',
    'فقحبة',
]

print('Profanity Filter — Test Cases:')
for t in test_cases:
    filtered = filter_profanity(t)
    changed = '✅ clean'
    if normalize_arabic(t) == filtered:
      changed = 'X filtered'
      print(t)

    if normalize_arabic(t) != filtered:
        print(f'          |  OUT: {filtered}')

print(f'\n✅ Profanity filter ready ({len(BAD_WORDS)} extracted entries)')
```
